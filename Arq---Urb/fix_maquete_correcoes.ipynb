# Notebook para corrigir erros de digitação e atualizar imagens no portfólio acadêmico do projeto 'Maquete de Conforto Ambiental e Paisagismo'
Este notebook inspeciona os arquivos do site, corrige caracteres incorretos e atualiza as referências do portfólio para as novas fotos do projeto.

```python
# Importar bibliotecas Python para manipular arquivos e texto
import re
from pathlib import Path

site_root = Path(r'\\srvconstser01\Planilha de Controles\KAYK\antigavty.py (codar e rodar sites)\sistema-controle-estoque-local.git\Arq---Urb')
index_path = site_root / 'index.html'
style_path = site_root / 'style.css'

print('Site root:', site_root)
print('index.html exists:', index_path.exists())
print('style.css exists:', style_path.exists())
```

```python
# Ler arquivos HTML/CSS do site e inspecionar conteúdo atual
with index_path.open('r', encoding='utf-8', errors='replace') as f:
    index_text = f.read()

with style_path.open('r', encoding='utf-8', errors='replace') as f:
    style_text = f.read()

print('index.html length:', len(index_text))
print('style.css length:', len(style_text))

for pattern in ['Ã', 'â', 'Â', 'â€“', 'Ã¡', 'Ã§', 'Ã£', 'Â°', 'Ãª']:
    if pattern in index_text:
        print('Found pattern in index.html:', pattern)

for pattern in ['Ã', 'â', 'Â']:
    if pattern in style_text:
        print('Found pattern in style.css:', pattern)
```

```python
# Detectar e corrigir erros de ortografia e caracteres corrompidos
replacements = {
    'â€“': '–',
    'â€”': '—',
    'â€': '”',
    'â€œ': '“',
    'Ã¡': 'á',
    'Ã©': 'é',
    'Ã­': 'í',
    'Ã³': 'ó',
    'Ãº': 'ú',
    'Ã£': 'ã',
    'Ãµ': 'õ',
    'Ãª': 'ê',
    'Ã¢': 'â',
    'Ã§': 'ç',
    'Ã ': 'à',
    'Ã¹': 'ù',
    'Ã¨': 'è',
    'Ãœ': 'Ü',
    'Ã ': 'À',
    'Ã‰': 'É',
    'Ã’': 'Ò',
    'Ã’': 'Ó',
    'Ã': 'Á',
    'Ã‰': 'É',
    'ÃŠ': 'Ê',
    'Ã‹': 'Ë',
    'ÃŽ': 'Î',
    'Ã': 'Ï',
    'Ã‘': 'Ñ',
    'Ã–': 'Ö',
    'Ãœ': 'Ü',
    'ÃŸ': 'ß',
    'Ã‰': 'É',
    'Ã—': '×',
    'Â': '',
}

for wrong, right in replacements.items():
    index_text = index_text.replace(wrong, right)
    style_text = style_text.replace(wrong, right)

print('Correções aplicadas. Verificando uso de sequências corrompidas...')
broken = sorted({p for p in ['Ã', 'â', 'Â'] if p in index_text})
print('Remaining broken in index.html:', broken)
```

```python
# Substituir imagens no portfólio do projeto 'Maquete de Conforto Ambiental e Paisagismo'
project_folder = 'trabalho-op/Maquete de Conforto Ambiental e Paisagismo'
gallery_block = re.search(r'<div class="gallery-grid-mini">(.*?)</div>', index_text, re.S)
if gallery_block:
    new_gallery = [
        f'<img src="{project_folder}/WhatsApp Image 2026-05-20 at 21.44.35.jpeg" alt="Maquete - Ângulo 1" />',
        f'<img src="{project_folder}/WhatsApp Image 2026-05-20 at 21.44.35 (1).jpeg" alt="Maquete - Ângulo 2" />',
        f'<img src="{project_folder}/WhatsApp Image 2026-05-20 at 21.44.37.jpeg" alt="Maquete - Ângulo 3" />',
        f'<img src="{project_folder}/WhatsApp Image 2026-06-01 at 16.03.44.jpeg" alt="Maquete - Detalhe" />',
    ]
    replacement = '<div class="gallery-grid-mini">
' + '
'.join(['          ' + line for line in new_gallery]) + '
        </div>'
    index_text = re.sub(r'<div class="gallery-grid-mini">.*?</div>', replacement, index_text, count=1, flags=re.S)
    print('Galeria atualizada no HTML.')
else:
    print('Bloco de galeria não encontrado.')
```

```python
# Validar alterações e salvar arquivos
with index_path.open('w', encoding='utf-8') as f:
    f.write(index_text)

with style_path.open('w', encoding='utf-8') as f:
    f.write(style_text)

print('Arquivos salvos com UTF-8.')
print('Verificando novamente se ainda há sequências corrompidas...')
for text, name in ((index_text, 'index.html'), (style_text, 'style.css')):
    found = [p for p in ['Ã', 'â', 'Â'] if p in text]
    print(f'{name}:', found)
```
